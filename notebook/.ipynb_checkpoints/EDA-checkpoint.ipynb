```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

fichier_1 = pd.read_csv("../data/raw/livraisons_10000.csv")
plt.figure(figsize=(10, 5))
plt.hist(fichier_1["duree_reelle_min"], bins=30, edgecolor="black")
plt.xlabel("La duréé en minute")
plt.ylabel("Le nombre de livraions")
plt.show
```

```python
plt.figure(figsize=(10, 5))
plt.hist(fichier_1["distance_km"], bins=30, edgecolor="black")
plt.xlabel("Distance en Km")
plt.ylabel("Nombre de livraison")
plt.show
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(fichier_1["distance_km"], fichier_1["duree_reelle_min"],edgecolor="black", alpha=0.5)
plt.title("La distance en km et la duree reelle en min")
plt.xlabel("La distance en Km")
plt.ylabel("La duree en min")
plt.show
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(fichier_1["poids_kg"], fichier_1["duree_reelle_min"], alpha=0.5, edgecolor="black")
plt.title("Le poids et la duree de la livraison")
plt.xlabel("Poids en km")
plt.ylabel("Duree reelle en min")
plt.show
```

```python
# 1. Calculer la matrice de corrélation
# On ne garde que les colonnes numériques pour éviter les erreurs
df_numerique = fichier_1.select_dtypes(include=['float64', 'int64'])
matrice_corr = df_numerique.corr()

# 2. Configurer la taille de la figure
plt.figure(figsize=(10, 8))

# 3. Créer la heatmap avec Seaborn
sns.heatmap(
    matrice_corr, 
    annot=True,       # Affiche les valeurs numériques dans les cases
    cmap='coolwarm',  # Palette de couleurs (bleu = négatif, rouge = positif)
    fmt=".2f",        # Arrondit les chiffres à 2 décimales
    linewidths=0.5,   # Ajoute de fines lignes entre les cases
    vmin=-1, vmax=1   # Fixe les limites de l'échelle de -1 à 1
)

# 4. Titre et affichage
plt.title("Matrice de corrélation des variables numériques", fontsize=14)
plt.show()  
```

```python
# B. Boxplot groupé par catégorie (Très utile !)
# Exemple : Voir la répartition des prix selon la catégorie du produit
plt.figure(figsize=(10, 5))
sns.boxplot(x='duree_reelle_min', y='type_vehicule', data=fichier_1, palette='Set2')

plt.title("Durée par type de véhicule")
plt.xticks(rotation=45) # Pivoter les étiquettes si elles sont trop longues
plt.show()
```

```python
plt.figure(figsize=(6, 5))

# Tracer le boxplot d'une seule colonne
sns.boxplot(y=fichier_1['duree_reelle_min'], color='orange')

plt.title("Boxplot - Détection des valeurs extrêmes")
plt.ylabel("Durée réelle (min)")
plt.show()
```
