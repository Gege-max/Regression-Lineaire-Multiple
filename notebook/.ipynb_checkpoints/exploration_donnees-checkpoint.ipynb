Exploration des donnees de 9975 lignes et 11 colonnes d'entrainement

```python
# Importation des bibliotheques
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from cleaning import nettoyer_donnees
```

```python
#Lecture du fichier avec pandas

data_frame = pd.read_csv("../data/livraisons_train_propre.csv")
print("Le nombre de ligne et colonnes sont :", data_frame.shape)
print("Le nombre de dimension est :", data_frame.ndim)
print("Le nombre de colonne est : ",data_frame.columns)
print("Type de donnee : ", data_frame.dtypes)
print("Information general : ", data_frame.info())
print("Verification des doublons : " ,data_frame.duplicated().sum())
```

Visualisation des données

```python
#Histogramme pour chaque variables numeriques

plt.figure(figsize=(10, 5))
plt.hist(data_frame["distance_km"], bins=30, edgecolor="black")
plt.title("La distance en km des livraisons")
plt.ylabel("Nombres de livraisons")
plt.show()
```

Je remarque que dans les données il y'a des distances qui sont inferieurs ou egale a 0 je vais faire un boxplots pour avoir encore plus de details


```python
plt.figure(figsize=(10, 5))
sns.boxplot(y=data_frame["distance_km"], color="red")
plt.title("Boxplot de la distance")
plt.ylabel("valeurs")
plt.show()
```

alors je vais le nettoyer dans mon notebook cleaning, ensuite appelé la fonction afin d'arranger le probleme de mon feature vu que c'est bien ce que je pensais

```python
df_nettoyage = nettoyer_donnees(data_frame)
print(df_nettoyage["distance_km"].head)
print(df_nettoyage["duree_reelle_min"].sum)
```

```python
plt.figure(figsize=(10, 5))
plt.hist(df_nettoyage["distance_km"], bins=30, edgecolor="black")
plt.title("Histogramme de la distance")
plt.ylabel("Les nombres de livraisons")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(y=df_nettoyage["distance_km"], color="red")
plt.title("Boxplot de la distance")
plt.ylabel("Valeurs")
plt.show()
```

```python
#Visualisation de notre variable cible avec un histogramme
plt.figure(figsize=(10, 5))
plt.hist(df_nettoyage["duree_reelle_min"], bins=30, edgecolor="black")
plt.title("La durée des livraisons en min")
```

```python
#Visualisation avec un diagramme en barre de des variables categoriques
ordre = ["lundi", "mardi", "mercredi", "jeudi", "vendredi", "samedi", "dimanche"] 
plt.figure(figsize=(10, 5))
sns.countplot(data=df_nettoyage["jour_semaine"], order=ordre)
plt.title("Jour de la semaine")
plt.xlabel("Nombres de livraison")
plt.show()
```

```python

plt.figure(figsize=(10, 5))
sns.countplot(data = df_nettoyage["meteo"])
plt.title("La meteo")
plt.xlabel("Nombres de livraison")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(df_nettoyage["type_vehicule"])
plt.title("Les types de vehicules")
plt.xlabel("Nombres de livraisons")
plt.show()
```

```python
#Calculer la matrice de correlation
numerique = df_nettoyage.select_dtypes(include=["int64", "float64"])
correlation_matrice= numerique.corr()
plt.figure(figsize=(8, 6))
# Afficher la matrice sous forme de Heatmap (carte de chaleur)
sns.heatmap(
    correlation_matrice, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)
plt.title("Matrice de correlation des variables", fontsize=14)
plt.show()
```

```python
#Visualidation avec les nuages des points la relation entre la distance et la cible
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df_nettoyage['distance_km'], y=df_nettoyage["duree_reelle_min"])
plt.title("Relation entre la distance et la durée réelle")
plt.xlabel("Distance (km)")
plt.ylabel("Durée réelle (min)")
plt.show()
```

```python
#Visualidation avec les nuages des points la relation entre la distance et la cible
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df_nettoyage['nb_arrets'], y=df_nettoyage["duree_reelle_min"])
plt.title("Relation entre la distance et la durée réelle")
plt.xlabel("Nombre d'arrets")
plt.ylabel("Durée réelle (min)")
plt.show()
```

```python
#Visualidation avec les nuages des points la relation entre la distance et la cible
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df_nettoyage["trafic_indice"], y=df_nettoyage["duree_reelle_min"])
plt.title("Relation entre la distance et la durée réelle")
plt.xlabel("Trafic d'indice")
plt.ylabel("Durée réelle (min)")
plt.show()
```
